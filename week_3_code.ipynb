## Task 1: Create the inventory and order queue

```python
# Task 1: Inventory dictionary and order queue
inventory = {"Widget": 10, "Gadget": 5, "Sensor": 0, "Cable": 15, "Bolt": 12}

orders = [
    ["Widget", 3],
    ["Sensor", 2],
    ["Gadget", 7],
    ["Cable", 15],
    ["Gizmo", 4],
    ["Widget", 8],
    ["Bolt", 8],
    ["Cable", 2],
]

print("Starting inventory:", inventory)
print("Orders in queue:")
for order in orders:
    print("  ", order)
```

## Task 2: Process the orders

```python
# Task 2: Process each order
unfulfilled = []       # items/quantities that could not be fully supplied
fully_fulfilled = 0    # counter for fully fulfilled orders

print("=== PROCESSING ORDERS ===")

for order in orders:
    item = order[0]
    qty = order[1]
    stock = inventory.get(item)  # returns None if the key does not exist

    if stock is None:
        print(f"ALERT: '{item}' does not exist in inventory. Order of {qty} rejected.")
        unfulfilled.append((item, qty))
    elif stock >= qty:
        inventory[item] = stock - qty
        fully_fulfilled += 1
        print(f"SUCCESS: {qty} x {item} fulfilled. Remaining stock: {inventory[item]}")
    elif stock > 0:
        remainder = qty - stock
        inventory[item] = 0
        unfulfilled.append((item, remainder))
        print(f"PARTIAL: {stock} x {item} shipped. {remainder} unfulfilled.")
    else:
        print(f"ALERT: '{item}' is out of stock. Order of {qty} not filled.")
        unfulfilled.append((item, qty))
```

## Task 3: Analytics and summary output

```python
# Task 3: Analytics and summary output
print("=== END-OF-BATCH SUMMARY ===")
print("Final inventory:", inventory)
print("Fully fulfilled orders:", fully_fulfilled)
print("Unfulfilled items (item, quantity):")
for entry in unfulfilled:
    print(f"  - {entry[0]}: {entry[1]}")
```
